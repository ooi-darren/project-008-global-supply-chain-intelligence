# 02 -- Historical Changes

## Research Question
What changed structurally in global supply chains since 2016 -- and which changes are temporary, cyclical, structural, strategic, technological, geopolitical, or environmental?

## Data
World Bank trade-openness panel, 2013-2025 (`data/processed/master_supply_chain_panel.csv`), PUBLIC. UN Comtrade network, 2016 vs. 2023, PUBLIC.

## Method
Median trade openness (% of GDP) across tracked economies by year, annotated against widely-documented global shock years (not a statistically-detected structural break -- a documented, honest limitation).

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import Image, display
import sys, os
sys.path.insert(0, os.path.abspath('../python/visualisation'))
pd.set_option('display.max_columns', 20)
```

```python
panel = pd.read_csv('../data/processed/master_supply_chain_panel.csv')
world_trend = panel.groupby('Year')['trade_pct_of_gdp'].median()
print(world_trend.round(1))
```

## Analysis

Visualisation 12 shows this trend with COVID (2020) and the Ukraine-war/energy-shock (2022) years marked.

```python
display(Image('../outputs/figures/12_structural_change_timeline.png', width=750))
```

## Finding

Median trade openness across the tracked panel does not show a dramatic, permanent break at either 2020 or 2022 -- consistent with the "supply chains proved more resilient in aggregate than the headlines suggested" narrative that has emerged in post-COVID trade literature, though this project's own data cannot establish causation, only the aggregate pattern.

The more genuinely structural finding in this project is the 2016-\>2023 network-centrality shift documented in Notebook 06: India, the Netherlands, Belgium, Poland, and the United States gaining real network importance, and the UK, Hong Kong SAR, and Japan losing it (Iran losing the most of all), is a more precise, harder-to-fake signal of structural change than an aggregate openness ratio.

## Limitation

Two time points (2016, 2023) for the network, not a full annual series -- see `docs/LIMITATIONS.md` item 3.
